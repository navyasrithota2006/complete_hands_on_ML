here in the above,date_of_journey is the object which is a date and date is not understandable for ml algo,so our task is to seperate the date to dat,month,year

```python
#feature engineering
df['Date'] = df['Date_of_Journey'].str.split('/').str[0] #only the day of all cols by splitting
df['Date']
```

```python
df['Month'] = df['Date_of_Journey'].str.split('/').str[1]
df['Month'] #gets the month
```

```python
df['Year'] = df['Date_of_Journey'].str.split('/').str[2]
df['Year']
```

```python
df.head()
```

```python
df.info()
```

here date,month,year are added and are in the type object,we need to change that

```python
df['Date'] = df['Date'].astype(int)
df['Month'] = df['Month'].astype(int)
df['Year'] = df['Year'].astype(int)
```

```python
df.info()
```

```python
#Drop Date of Journey
df.drop('Date_of_Journey',axis = 1,inplace=True)
```

```python
df.head()
```

```python
df['Arrival_hour'] = df['Arrival_Time'].str.split(':').str[0]
df['Arrival_minute'] = df['Arrival_Time'].str.split(':').str[1]
```

```python
df.head()
```

here we are getting the arrival minute with al the '22march" or something like that and i dont want that so i use lambda function 

```python
df['Arrival_Time'] = df['Arrival_Time'].apply(lambda x:x.split(' ')[0])
```

```python
df['Arrival_hour'] = df['Arrival_Time'].str.split(':').str[0]
df['Arrival_minute'] = df['Arrival_Time'].str.split(':').str[1]
```

```python
df.drop('Arrival_Time',axis=1,inplace=True)
```
