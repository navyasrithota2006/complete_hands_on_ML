```python
import sqlite3
```

```python
#working with the sales data
#connect to the sqlite database
connection = sqlite3.connect('sales_data.db')
connection
```

```python
cursor = connection.cursor()
```

```python
#create table for the sales data
cursor.execute('''
create table if not exists sales(
id integer primary key,
data text not null,
product text not null,
sales integer,
region text)
''')
connection.commit()
```

```python
#insert values

sales_data = [
    ('2023-01-01','product1',100,'north'),
    ('2023-01-02','product2',200,'south'),
    ('2023-01-03','product3',300,'east'),
    ('2023-01-04','product4',400,'west'),
    ('2023-01-05','product5',500,'north'),
]
```

```python
#we can insert bulk data int tthe table 

cursor.executemany('''
insert into sales(data,product,sales,region) 
values(?,?,?,?)
''',sales_data)
connection.commit()
```

```python
cursor.execute('select * from sales')
rows = cursor.fetchall()
for row in rows:
    print(row)
```

```python
connection.close()
```
