###sqlite

self contained,serverless,zero configuartion database engine that is widely used for embedded database engine

```python
import sqlite3
```

```python
#connect to an sqlite database

connection = sqlite3.connect('example.db')
connection
```

```python
#we have to create a cursor object which iterates through tables of rows and columns

cursor = connection.cursor()
```

```python
#create a table using cursor.execute

cursor.execute('''
create table if not exists employees(
if integer primary key,
name text not null,
age integer,
department text)
''')

#commit the changes
connection.commit()
```

```python
#insert the data into the sqlite table
cursor.execute('''
insert into employees(name,age,department) values('krish',32,'data enginner')
''')
cursor.execute('''
insert into employees(name,age,department) values('navya',23,'ml enginner')
''')
cursor.execute('''
insert into employees(name,age,department) values('chandrika',32,'doctor')
''')

connection.commit()
```

```python
#query the data from the table

cursor.execute('select * from employees')
rows = cursor.fetchall()
for row in rows:
    print(row)
```

```python
#update data in the table

cursor.execute('''
update employees set age = 32 where name = 'navya'
''')

connection.commit()
```

```python
#delete the data from table

cursor.execute('''
delete from employees where name = 'navya'

''')

connection.commit()
```

```python
#close the connection

connection.close()
```
