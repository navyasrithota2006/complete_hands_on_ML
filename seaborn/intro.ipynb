```python
import seaborn as sns
```

```python
##basic plotting with seaborn
tips = sns.load_dataset('tips') #there are built in datasets and we can just load them
tips
```

```python
##create a scatter plot
import matplotlib.pyplot as plt # normally we use matplotlib for scatterplot

sns.scatterplot(x = 'total_bill',y='tip',data = tips) #x,y are the column names and data is the whole data
plt.title("scatter plot of total bill vs tip ")
plt.show()
```

```python
##Line plot

sns.lineplot(x='size',y='total_bill',data = tips)
plt.show()
```

```python
##categorical plots
##bar plot

sns.barplot(x = 'day',y='total_bill',data = tips)
plt.title('bar plot of total bill by day')
plt.show()
```

```python
#box plot
sns.boxplot(x='day',y='total_bill',data = tips)
```

```python
##violin plot

sns.violinplot(x = 'day',y='total_bill',data = tips)
```

```python
##histograms

sns.histplot(tips['total_bill'],bins = 10,kde=True) #kde shows how the data is being actually distributed
```

```python
##KDE plot (kernel density estimator)
sns.kdeplot(tips['total_bill'],fill = True)
```

```python
#pairplot - if i want to see multiple variables relations
sns.pairplot(tips)
```

```python
##heatmap uses correlation matrix
#correlation tells us how strongly two numerical variables are related to each other
#df.corr will give the dataframe from -1 to 1 values
corr = tips[['total_bill','tip','size']].corr()
corr
```

```python
sns.heatmap(corr)
```

```python
sns.heatmap(corr,annot=True,cmap='coolwarm')  # annot to display
```

```python
import pandas as pd
sales_df = pd.read_csv('sales.csv',encoding='cp1252')
sales_df.head() 
```

```python
##plot total sales per year
plt.figure(figsize=(10,6))
sns.barplot(x = "YEAR_ID",y='SALES',data=sales_df,estimator=sum) #estimator is used for aggregation functions
plt.title("total sales per year")
plt.xlabel("year")
plt.ylabel("total sales")
plt.show()
```

```python
##plot total sales per month
plt.figure(figsize=(10,6))
sns.barplot(x = "MONTH_ID",y='SALES',data=sales_df,estimator=sum) #estimator is used for aggregation functions
plt.title("total sales per MONTH")
plt.xlabel("MONTH")
plt.ylabel("total sales")
plt.show()
```
