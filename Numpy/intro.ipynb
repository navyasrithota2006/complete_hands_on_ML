**np.arange**

create an ndarray with given range

```python
print(np.arange(1,5))
print(np.arange(1.0,5.0))
print(np.arange(1,5,0.5))
```

```python
#we cant always set up floats because elements in array is not predictable
print(np.arange(0,5/3,1/3))
print(np.arange(0,5/3,0.3333333))
print(np.arange(0,5/3,0.3333334)) #just by increasing one digit,the elements are lesser
```

**np.linspace**

for the above problem when using floats,we use linspace function  which returns a array with specific number of points evenly distributed between two values

**np.random**

is used to create the ndarrays with random values

before doing this ,we need to create a random number generator like np.random.default_rng(seed),seed is given to produce reproducible results

```python
rng = np.random.default_rng(seed = 42)
```

```python
rng.random((3,4)) #random floats between 0 and 1 (uniform distribution)
```

```python
rng.standard_normal((3,4)) #gaussian distribution or normal distribution of mean 0 and variance 1
```

```python
rng.integers(10,20,(3,4)) #random integers from 10 to 19 in 3* 4 matrix
```

```python
rng.permutation(10) # random permutation of integers from 0 to 9
```

to view the distribution ,we see the matplotlib

```python
import matplotlib.pyplot as plt
```

```python
plt.hist(rng.random(100000), density=True, bins=100, histtype="step", color="blue", label="rand")
plt.hist(rng.standard_normal(100000), density=True, bins=100, histtype="step", color="red", label="randn")
plt.axis([-2.5, 2.5, 0, 1.1])
plt.legend(loc = "upper left")
plt.title("Random distributions")
plt.xlabel("Value")
plt.ylabel("Density")
plt.show()
```
