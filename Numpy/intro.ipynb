Numpy


numpy is mainly centered around n-dimensional array object
it also contains linear algebra,fourier transform,random number functions

```python
#importing the numpy
import numpy as np
```

```python
#np.zeros - creates an array containing number of zeroes
np.zeros(5)
```

```python
np.zeros((3,4))
#takes tuple as number of rows and columns
```

In numpy, dimension = axis

number of axes = rank (eg : 3*4 matrix is a 2d array of rank 2,the length of 1st axis=3 and 2nd = 4)

list of axis is called shape.like (3,4).rank = shape's length

size of array = total number of elements = product of axis length 

size of array = 12 = 3*4 =12


```python
a = np.zeros((3,4))
a
```

np.ones -creates array with all ones

**np.full**

creates the array with full of given values with given shape

```python
np.full((3,4),np.pi)
```

**np.empty**

it is uninitialised with whatever memory has at that point with the given shape

```python
np.empty((2,3))
```

**np.array**

initialise an array using python array

```python
np.array([[1,2,3,4],[10,202,30,40]])
```

**np.arange**

create an ndarray with given range

```python
print(np.arange(1,5))
print(np.arange(1.0,5.0))
print(np.arange(1,5,0.5))
```

```python
#we cant always set up floats because elements in array is not predictable
print(np.arange(0,5/3,1/3))
print(np.arange(0,5/3,0.3333333))
print(np.arange(0,5/3,0.3333334)) #just by increasing one digit,the elements are lesser
```

**np.linspace**

for the above problem when using floats,we use linspace function  which returns a array with specific number of points evenly distributed between two values

```python
np.linspace(0,5/3,6)
```
