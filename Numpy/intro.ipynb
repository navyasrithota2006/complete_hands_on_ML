**np.random**

is used to create the ndarrays with random values

before doing this ,we need to create a random number generator like np.random.default_rng(seed),seed is given to produce reproducible results

```python
rng = np.random.default_rng(seed = 42)
```

```python
rng.random((3,4)) #random floats between 0 and 1 (uniform distribution)
```

```python
rng.standard_normal((3,4)) #gaussian distribution or normal distribution of mean 0 and variance 1
```

```python
rng.integers(10,20,(3,4)) #random integers from 10 to 19 in 3* 4 matrix
```

```python
rng.permutation(10) # random permutation of integers from 0 to 9
```

to view the distribution ,we see the matplotlib

```python
import matplotlib.pyplot as plt
```

```python
plt.hist(rng.random(100000), density=True, bins=100, histtype="step", color="blue", label="rand")
plt.hist(rng.standard_normal(100000), density=True, bins=100, histtype="step", color="red", label="randn")
plt.axis([-2.5, 2.5, 0, 1.1])
plt.legend(loc = "upper left")
plt.title("Random distributions")
plt.xlabel("Value")
plt.ylabel("Density")
plt.show()
```

**np.fromfunction**

we could create ndarrays using the function in python

```python
def my_function(z,y,x):
    return x + 10 * y + 100 *z

np.fromfunction(my_function,(3,2,10))
```

numpy creates 3 ndarrays with each of the array is equal to their co ordinates along the specific axis(0,1,2).

here function is called only once not for every element.this makes initialization efficient

***Array Data***

**dtype**

numpy ndarrays have same elements.we can check the type of the elements in the arrays

```python
c= np.arange(1,5)
print(c.dtype)
print(c.dtype,c)
```

```python
#instead of letting numpy guees the dtype,we can manually set it
d = np.arange(1,5,dtype=np.complex64)
print(d.dtype,d)
```

available datatype are signed int,int32,int64,unsigned uint8|16|32|64 and complex64|128,float16|32|64

**itemsize**

attribute returns the size(in bytes) of each item

```python
e = np.arange(1,5,dtype = np.complex64)
print(e)
print(e.itemsize)
```
