```python
# Assignment 2: DataFrame Operations

#1. Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers.
#  Add a new column that is the product of the first two columns.

df = pd.DataFrame(np.random.randint(1,100,size=(5,3)),columns = ['X','Y','A'])
print("original array")
print(df)

df['Z'] = df["X"] * df['Y']
print("updated new column")
print(df)
```

```python
#2. Create a Pandas DataFrame with 3 columns and 4 rows filled with random integers. 
# Compute the row-wise and column-wise sum.

df = pd.DataFrame(np.random.randint(1,99,size = (4,3)),columns = ['A','B','C'])
print("original dataframe")
print(df)

row_sum = df.sum(axis =1)
col_sum = df.sum(axis =0)

print("row sum")
print(row_sum)
print("col sum")
print(col_sum)
```

```python
# Assignment 3: Data Cleaning

#1. Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. 
# Introduce some NaN values. Fill the NaN values with the mean of the respective columns.
df = pd.DataFrame(np.random.randint(1,100,size=(5,3)),columns = ['A','B','C'])
print("original data frame")
print(df)

df.iloc[0,1] = np.nan  # it initiales in 0th row ,1st col
df.iloc[2,2] = np.nan
df.iloc[4,0] = np.nan
print("data frame with nan values")
print(df)

df.fillna(df.mean(),inplace = True)
print("Data frame with nan values filled :")
print(df)
```

```python
#2. Create a Pandas DataFrame with 4 columns and 6 rows filled with random integers.
#  Introduce some NaN values. 
# Drop the rows with any NaN values.

df = pd.DataFrame(np.random.randint(1,100,size=(6,4)),columns = ['A','B','C','D'])
print("original data frame")
print(df)

df.iloc[0,1] = np.nan
df.iloc[2,3] = np.nan
print("datframe with nan values")
print(df)

df.dropna(inplace = True)
print("data frame after droping")
print(df)
```

```python
### Assignment 4: Data Aggregation

#1. Create a Pandas DataFrame with 2 columns: 'Category' and 'Value'. Fill the 'Category' column with random categories ('A', 'B', 'C') and the 'Value' column with random integers. 
# Group the DataFrame by 'Category' and compute the sum and mean of 'Value' for each category.

df = pd.DataFrame({'category':np.random.choice(['A','B','C'],size = 10),'value':np.random.randint(1,100,size=10)})
print("oroginal dataframe")
print(df)

grouped = df.groupby('category')['value'].agg(['sum','mean'])
print("grouped dataframe")
print(grouped)
```

```python
#2. Create a Pandas DataFrame with 3 columns: 'Product', 'Category', and 'Sales'. 
# Fill the DataFrame with random data.
#  Group the DataFrame by 'Category' and compute the total sales for each category.

df = pd.DataFrame({'product':np.random.choice(['prod1','prod2','prod3'],size=10),'category':np.random.choice(['A','B','C'],size = 10),'sales': np.random.randint(1,100,size=10)})
print("original array")
print(df)

grouped = df.groupby('category')['sales'].sum()
print("grouped dataframe")
print(grouped)
```

```python
### Assignment 5: Merging DataFrames

#1. Create two Pandas DataFrames with a common column.
#  Merge the DataFrames using the common column.

df1 = pd.DataFrame({'key':['A','B','C','D'],'value': np.random.randint(1,20,size = 4)})
df2 = pd.DataFrame({'key':['A','B','C','E'],'value': np.random.randint(1,20,size = 4)})
print("dataframe 1")
print(df1)
print("datframe 2")
print(df2)

#merging
merged = pd.merge(df1,df2,on='key')
print("merged dataframe")
print(merged)
```
