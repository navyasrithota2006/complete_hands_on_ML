```python
### Assignment 4: Statistical Operations

#1. Create a NumPy array of shape (5, 5) filled with random integers. Compute the mean, median, standard deviation, and variance of the array.
arr12 = np.random.randint(1,21,size=(5,5))
print(arr12)
#mean
mean = np.mean(arr12)
print("mean")
print(mean)
#median
median=np.median(arr12)
print("median")
print(median)
#standard deviation
std=np.std(arr12)
print("standard deviation")
print(std)
#variance
var = np.var(arr12)
print("variance")
print(var)
```

```python
#2. Create a NumPy array of shape (3, 3) with values from 1 to 9.
#  Normalize the array (i.e., scale the values to have a mean of 0 and a standard deviation of 1).
arr13 = np.random.randint(1,10,size=(3,3))
print(arr13)
mean = np.mean(arr13)
std = np.std(arr13)
normalised_array = (arr13 - mean)/std
print(normalised_array)
```

```python
### Assignment 5: Broadcasting

#1. Create a NumPy array of shape (3, 3) filled with random integers.
#  Add a 1D array of shape (3,) to each row of the 2D array using broadcasting.

arr14= np.random.randint(1,21,size=(3,3))
print(arr14)
oned_array = np.random.randint(1,21,size=(3,))
print(oned_array)
result = arr14 + oned_array
print(result)
```

```python
#2. Create a NumPy array of shape (4, 4) filled with random integers.
#  Subtract a 1D array of shape (4,) from each column of the 2D array using broadcasting.

arr15 = np.random.randint(1,16,size=(4,4))
print(arr15)
print("1d array")
onedim = np.random.randint(1,21,size=(4,))
print(onedim)
result =  arr15 - onedim[:,np.newaxis]
print("result")
print(result)
```

```python
# Assignment 6: Linear Algebra

#1. Create a NumPy array of shape (3, 3) representing a matrix. Compute its determinant, inverse, and eigenvalues.
arr16 = np.random.randint(1,10,size=(3,3))
print(arr16)
print("determinant")
print(np.linalg.det(arr16))
print("inverse")
print(np.linalg.inv(arr16))
print("eigenvalues")
print(np.linalg.eigvals(arr16))
```

```python
# Assignment 8: Fancy Indexing and Boolean Indexing

#1. Create a NumPy array of shape (5, 5) filled with random integers.
#  Use fancy indexing to extract the elements at the corners of the array.

arr21 = np.random.randint(1,21,size=(5,5))
print("original array")
print(arr21)
print("extracting the corner elements")
print(arr21[[0,0,-1,-1],[0,-1,0,-1]]) #(0,0),(0,-1),(-1,0),(-1,-1)
```

```python
#2. Create a NumPy array of shape (4, 4) filled with random integers. Use boolean indexing to set all elements greater than 10 to 10.

arr22 = np.random.randint(1,17,size=(4,4))
print("original array")
print(arr22)
arr22[arr22 > 10] = 10
print("modified array")
print(arr22)
```

```python
# Assignment 9: Structured Arrays

#1. Create a structured array with fields 'name' (string), 'age' (integer), and 'weight' (float). 
# Add some data and sort the array by age.

data_type = [('name','U10'),('age','i4'),('weight','f4')]
data = np.array([('Alice',15,65.5),('bob',13,54.3),('charlie',34,43.6)],dtype=data_type)
print("original array")
print(data)

#sort the data by age
sorted_data = np.sort(data,order='age')
print("sorted by age")
print(sorted_data)
```

```python
#2. Create a structured array with fields 'x' and 'y' (both integers).
#  Add some data and compute the Euclidean distance between each pair of points.
data_type = [('x','i4'),('y','i4')]
data = np.array([(1,2),(3,4),(5,6)],dtype = data_type)
print("original array")
print(data)
distances = np.sqrt((data['x'][:,np.newaxis] - data['x'])**2 + (data['y'][:,np.newaxis] - data['y'])**2)
print("euclidean distances")
print(distances)
```

```python
## Assignment 10: Masked Arrays

#1. Create a masked array of shape (4, 4) with random integers and
#  mask the elements greater than 10. Compute the sum of the unmasked elements.

import numpy.ma as ma

arr23 = np.random.randint(1,17,size = (4,4))
print("original array")
print(arr23)

masked_array = ma.masked_greater(arr23,10)
print("masked array")
print(masked_array)

sum_masked_elements = masked_array.sum()
print("sum")
print(sum_masked_elements)
```

```python
#2. Create a masked array of shape (3, 3) with random integers and mask the diagonal elements.
#  Replace the masked elements with the mean of the unmasked elements.

arr24 = np.random.randint(1,10,size =(3,3))
print("original array")
print(arr24)
mas_array =ma.masked_array(arr24,mask = np.eye(3,dtype=bool))
print("masked array")
print(mas_array)

mean_of = arr24.mean()
print("mean: ",mean_of)
replaced_array = mas_array.filled(mean_of)
print("replacing array with the mean")
print(replaced_array)
```
