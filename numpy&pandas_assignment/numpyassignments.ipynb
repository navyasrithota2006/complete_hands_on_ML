```python
#Assignment 3: Array Operations

#1. Create two NumPy arrays of shape (3, 4) filled with random integers. Perform element-wise addition, subtraction, multiplication, and division.

arr5 = np.random.randint(1,21,size=(3,4))
print(arr5)
arr6 = np.random.randint(21,41,size=(3,4))
print(arr6)
#addition 
arr7 = arr5 + arr6
print(arr7)
#subtraction
arr8 = arr5 - arr6
print(arr8)
#multiplication
arr9 = arr5 * arr6
print(arr9)
#division
arr10 = arr5 /arr6
print(arr10)
```

```python
#2. Create a NumPy array of shape (4, 4) with values from 1 to 16. Compute the row-wise and column-wise sum.
arr11 = np.random.randint(1,17,size=(4,4))
print(arr11)
row_sum = np.sum(arr11,axis=1)
col_sum = np.sum(arr11,axis=0)
print("row_sum")
print(row_sum)
print("col_num")
print(col_sum)
```

```python
### Assignment 4: Statistical Operations

#1. Create a NumPy array of shape (5, 5) filled with random integers. Compute the mean, median, standard deviation, and variance of the array.
arr12 = np.random.randint(1,21,size=(5,5))
print(arr12)
#mean
mean = np.mean(arr12)
print("mean")
print(mean)
#median
median=np.median(arr12)
print("median")
print(median)
#standard deviation
std=np.std(arr12)
print("standard deviation")
print(std)
#variance
var = np.var(arr12)
print("variance")
print(var)
```

```python
#2. Create a NumPy array of shape (3, 3) with values from 1 to 9.
#  Normalize the array (i.e., scale the values to have a mean of 0 and a standard deviation of 1).
arr13 = np.random.randint(1,10,size=(3,3))
print(arr13)
mean = np.mean(arr13)
std = np.std(arr13)
normalised_array = (arr13 - mean)/std
print(normalised_array)
```

```python
### Assignment 5: Broadcasting

#1. Create a NumPy array of shape (3, 3) filled with random integers.
#  Add a 1D array of shape (3,) to each row of the 2D array using broadcasting.

arr14= np.random.randint(1,21,size=(3,3))
print(arr14)
oned_array = np.random.randint(1,21,size=(3,))
print(oned_array)
result = arr14 + oned_array
print(result)
```

```python
#2. Create a NumPy array of shape (4, 4) filled with random integers.
#  Subtract a 1D array of shape (4,) from each column of the 2D array using broadcasting.

arr15 = np.random.randint(1,16,size=(4,4))
print(arr15)
print("1d array")
onedim = np.random.randint(1,21,size=(4,))
print(onedim)
result =  arr15 - onedim[:,np.newaxis]
print("result")
print(result)
```
