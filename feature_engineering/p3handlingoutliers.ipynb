###handling outliers###

1. by using 5 number summary , we can find the outliers and  remove them

2. 5 number are
* minimum
* 1st quartile(25%)
* median
* 3rd quartile(75%)
* maximum

```python
import numpy as np
```

```python
lst_marks = [45,32,56,75,89,60,87,54,74,45,98,67,74,90]
minimum,Q1,median,Q3,maximum=np.quantile(lst_marks,[0,0.25,0.50,0.75,1.0])
```

```python
minimum,Q1,median,Q3,maximum
```

we have to define the lower fence and upper fence so that anything below or higher than lower fence and upper fence 

the lower fence formula is Q1- 1.5*IQR
upper fence formula = Q3 + 1.5*IQR
where IQR = Q3 - Q1

```python
IQR = Q3-Q1
print(IQR)
```

```python
#lower fence
lower_fence = Q1 - 1.5*IQR
upper_fence = Q3 + 1.5*IQR
```
