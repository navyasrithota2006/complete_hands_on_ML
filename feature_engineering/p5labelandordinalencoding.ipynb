### Label Encoding

assigns a new unique numerical values to each category in the column
 eg: if a column of 'colors' has 3 different colors with red,green,blue .so ny label encoding it is represented as 
 ** red : 1,
 ** green : 2,
 ** blue : 3

 it assigns unique numbers to each category 

```python
import pandas as pd
```

```python
#create a dataframe

df = pd.DataFrame({
    'color':['red','blue','green','green','red','blue']
})
```

```python
from sklearn.preprocessing import LabelEncoder
lbl_encoder = LabelEncoder()
```

```python
lbl_encoder.fit_transform(df[['color']])
```

```python
#get any value from the array after fit and transform
lbl_encoder.transform([['red']])
```

```python
lbl_encoder.transform([['blue']])
```

## disadvantage

as we are assigning the new values after training the ml model thinks that number of greens are greater than reds or numbers of blues are less than reds.

## optimisation : ordinal encoding

to not think and specifically need to reassign the ranks.so, we use ordinal encoding

## ordinal encoding

* encodes the categorical data with a ranking.
* in this, each category assigned a numerical value based on its position in the order.
eg :
 we have column name with "education level" with 4 possible values(high school,college,graduate,post-graduate)
 * the encoding is as follows 
  1. high school : 1
  2. college : 2
  3. graduate : 3
  4. pg : 4

```python
#ordinal encoding
from sklearn.preprocessing import OrdinalEncoder
```

```python
#create a datadrame
df = pd.DataFrame({
    'size' : ['small','medium','large','medium','small','large']
})
```

```python
df
```

```python
#create an instance for ordinal encoder
encoder = OrdinalEncoder(categories=[['small','medium','large']])
```

```python
encoder.fit_transform(df[['size']])
```

```python
encoder.transform([['small']])
```
