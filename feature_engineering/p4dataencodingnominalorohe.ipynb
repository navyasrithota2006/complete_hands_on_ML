### Data Encoding

process of converting categorical data(text or labels) to numerical values so that ml algorithms can understand and process it


### types of encoding

1. Nominal/OHE encoding
2. label and ordinal encoding
3. target guided ordinal encoding

### Nominal/One hot encoding

in this each category is represented as binary numbers with a unique category

eg: we have category like color which has different samples with green,blue,red and we want in one sample it creates green,blue,red features and initialise them with 1,0,0 if the samples says it has green in it

## disadvantage
1. it created n numbers of features with all the n samples.
2. it creates a sparse matrix(samples combined with 1's and 0's as matrix) which leads to overfitting(trains very well on trainning data but not on test data)

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder 
```

```python
#create a dataframe

df = pd.DataFrame({
    'color':['red','blue','green','green','red','blue']
})
```

```python
df
```

```python
#we create an instance for onehotencoder

encoder = OneHotEncoder()
```

```python
#perform and fit and transform
#here fit does is learn the required parameters from the data
#transform does is applies it to the data 
 
encoded = encoder.fit_transform(df[['color']]).toarray()
encoded
# the below arrays is formed alphabetical order
```

```python
import pandas as pd
encoder_df = pd.DataFrame(encoded,columns = encoder.get_feature_names_out())
```

```python
encoder_df
```

```python
encoder.transform([['blue']]).toarray()
```

```python
pd.concat([df,encoder_df],axis = 1)
```
