### Target guided ordinal encoding

* this is useful when we have a column with large number of unique categories

* we replace the category into numerical value based on 'mean or median' of the target coulumn for that category.

* this creates a monotonic relation ship between categorical column and the target column.

* this can improve predictive power of the ml model


```python
import pandas as pd
df = pd.DataFrame({
    'city' : ['New York','London','paris','tokyo','New York','paris'],
    'price' : [200,150,300,250,180,320]
})
```

```python
df
```

here we have 2 new york with price 200 & 180 ,now take a mean of it and assign new york with it


```python
mean_price = df.groupby('city')['price'].mean().to_dict()
```

```python
mean_price
```

```python
df['city_encoded'] = df['city'].map(mean_price)
```

when we map the city with mean price ,whenever there new york it is assign the mean of it other and kept in a new feature called city encoded

```python
df
```

```python
# for the model to train i can give only dataframe of price and the city encoded
df[['price','city_encoded']]
```
