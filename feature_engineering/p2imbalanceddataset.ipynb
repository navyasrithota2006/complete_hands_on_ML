### handling imbalanced dataset

1. Up sampling
2. down sampling

```python
import numpy as np
import pandas as pd
```

```python
np.random.seed(123)
n_samples = 1000
class_0_ratio = 0.9
n_class_0 = int(n_samples * class_0_ratio)
n_class_1 = n_samples - n_class_0
```

```python
n_class_0,n_class_1
```

```python
class_0 = pd.DataFrame({
    'feature_1' : np.random.normal(loc=0,scale=1,size = n_class_0),
    'feature_2' : np.random.normal(loc=0,scale=1,size=n_class_0),
    'target' : [0]*n_class_0
})
#hre loc = mean,scale = standard deviation,size = number of rows
class_1 = pd.DataFrame({
    'feature_1' : np.random.normal(loc =2,scale = 1,size = n_class_1),
    'feature_2' : np.random.normal(loc=2,scale = 1,size= n_class_1),
    'target' : [1]*n_class_1
})
```

```python
df=pd.concat([class_0,class_1]).reset_index(drop=True) #reset index is used when we concatenate the index repeat from 0 to n and again 0 to n,so to have continuous indexes we say
#reset index where drop is true
df
```

```python
df.tail()
```

```python
df['target'].value_counts()
```

```python
df_minority = df[df['target'] == 1]
df_majority = df[df['target'] == 0]
```

```python
from sklearn.utils import resample

df_minority_upsampled = resample(df_minority,replace = True,
                            n_samples = len(df_majority),
                            random_state = 42)
```

```python
df_minority_upsampled.shape
```

```python
df_minority_upsampled.head()
```

```python
df_upsampled = pd.concat([df_majority,df_minority_upsampled])
df_upsampled.shape
```

```python
df_upsampled['target'].value_counts()
```

###down sampling

```python
import pandas as pd

# Set the random seed for reproducibility
np.random.seed(123)

# Create a dataframe with two classes
n_samples = 1000
class_0_ratio = 0.9
n_class_0 = int(n_samples * class_0_ratio)
n_class_1 = n_samples - n_class_0

class_0 = pd.DataFrame({
    'feature_1': np.random.normal(loc=0, scale=1, size=n_class_0),
    'feature_2': np.random.normal(loc=0, scale=1, size=n_class_0),
    'target': [0] * n_class_0
})

class_1 = pd.DataFrame({
    'feature_1': np.random.normal(loc=2, scale=1, size=n_class_1),
    'feature_2': np.random.normal(loc=2, scale=1, size=n_class_1),
    'target': [1] * n_class_1
})

df = pd.concat([class_0, class_1]).reset_index(drop=True)

# Check the class distribution
print(df['target'].value_counts())
```

```python
df_minority = df[df['target']== 1]
df_majority = df[df['target'] == 0]
```

```python
from sklearn.utils import resample
df_majority_downsampled = resample(df_minority,replace=True,
                                n_samples = len(df_majority),
                                random_state = 42)
```

```python
df_majority_downsampled.shape
```

```python
df_downsamle = pd.concat([df_minority,df_majority_downsampled])
```
